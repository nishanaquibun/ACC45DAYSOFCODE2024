```python
import requests
import pandas as pd
import time

sensor_id = 12236153

url = f"https://api.openaq.org/v3/sensors/{sensor_id}/hours"

headers = {
    "X-API-Key": API_KEY
}

all_records = []

page = 1
limit = 1000

while True:

    params = {
        "date_from": "2025-02-19T00:00:00Z",
        "date_to": "2026-09-17T23:59:59Z",
        "limit": limit,
        "page": page
    }

    response = requests.get(
        url,
        headers=headers,
        params=params
    )

    print("Page:", page, "| Status:", response.status_code)

    if response.status_code != 200:
        print(response.text)
        break

    result = response.json()

    records = result.get("results", [])

    if not records:
        break

    all_records.extend(records)

    print("Records collected:", len(all_records))

    if len(records) < limit:
        break

    page += 1

    time.sleep(0.2)

print("\nTotal records:", len(all_records))
```

```python
rows = []

for record in all_records:

    rows.append({
        "datetime": record["period"]["datetimeFrom"]["local"],
        "pm25": record["value"]
    })

pm25_df = pd.DataFrame(rows)

pm25_df["datetime"] = pd.to_datetime(pm25_df["datetime"])

pm25_df = pm25_df.sort_values("datetime")

pm25_df = pm25_df.drop_duplicates(
    subset="datetime"
)

print(pm25_df.head())
print(pm25_df.tail())

print("\nTotal rows:", len(pm25_df))
```

```python
# Remove previously created feature columns
old_features = [
    "hour",
    "day",
    "day_of_week",
    "month",
    "is_weekend",
    "pm25_lag_1",
    "pm25_lag_3",
    "pm25_lag_6",
    "pm25_lag_12",
    "pm25_lag_24",
    "pm25_lag_48",
    "pm25_rolling_3h",
    "pm25_rolling_6h",
    "pm25_rolling_24h",
    "target_pm25"
]

df_model = df_model.drop(
    columns=[col for col in old_features if col in df_model.columns]
)

print(df_model.shape)
print(df_model.columns.tolist())
```

```python
print(df_model.head())
print("\nMissing values:")
print(df_model.isna().sum())
```

```python
# Time-based features

df_model["hour"] = df_model.index.hour
df_model["day"] = df_model.index.day
df_model["day_of_week"] = df_model.index.dayofweek
df_model["month"] = df_model.index.month

# Weekend flag
df_model["is_weekend"] = (
    df_model["day_of_week"] >= 5
).astype(int)

print(df_model.head())
print("\nNew shape:", df_model.shape)
```

```python
# PM2.5 lag features

lag_hours = [1, 3, 6, 12, 24, 48]

for lag in lag_hours:
    df_model[f"pm25_lag_{lag}"] = df_model["pm25"].shift(lag)

print(df_model[
    [
        "pm25",
        "pm25_lag_1",
        "pm25_lag_3",
        "pm25_lag_6",
        "pm25_lag_12",
        "pm25_lag_24",
        "pm25_lag_48"
    ]
].head(55))
```

```python
# Rolling PM2.5 features
# Shift by 1 hour to use only past PM2.5 values

df_model["pm25_rolling_3h"] = (
    df_model["pm25"]
    .shift(1)
    .rolling(window=3, min_periods=3)
    .mean()
)

df_model["pm25_rolling_6h"] = (
    df_model["pm25"]
    .shift(1)
    .rolling(window=6, min_periods=6)
    .mean()
)

df_model["pm25_rolling_24h"] = (
    df_model["pm25"]
    .shift(1)
    .rolling(window=24, min_periods=24)
    .mean()
)

print(
    df_model[
        [
            "pm25",
            "pm25_rolling_3h",
            "pm25_rolling_6h",
            "pm25_rolling_24h"
        ]
    ].head(30)
)
```

```python
import numpy as np

# Convert wind direction into circular features
df_model["wind_direction_sin"] = np.sin(
    np.deg2rad(df_model["wind_direction"])
)

df_model["wind_direction_cos"] = np.cos(
    np.deg2rad(df_model["wind_direction"])
)

# Target: PM2.5 one hour into the future
df_model["target_pm25"] = df_model["pm25"].shift(-1)

print(
    df_model[
        [
            "wind_direction",
            "wind_direction_sin",
            "wind_direction_cos",
            "pm25",
            "target_pm25"
        ]
    ].head(10)
)
```

```python
# Check missing values in our modeling features

print("Missing values:")
print(df_model.isna().sum())

print("\nTotal rows:", len(df_model))

# Create a copy for modeling
df_ml = df_model.dropna().copy()

print("\nRows available for modeling:", len(df_ml))
print("Rows removed:", len(df_model) - len(df_ml))
```

```python
# Final feature columns

features = [
    "temperature",
    "humidity",
    "pressure",
    "wind_speed",
    "rain",

    "hour",
    "day_of_week",
    "month",
    "is_weekend",

    "wind_direction_sin",
    "wind_direction_cos",

    "pm25_lag_1",
    "pm25_lag_3",
    "pm25_lag_6",
    "pm25_lag_12",
    "pm25_lag_24",
    "pm25_lag_48",

    "pm25_rolling_3h",
    "pm25_rolling_6h",
    "pm25_rolling_24h"
]

target = "target_pm25"

X = df_ml[features]
y = df_ml[target]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())
```

```python
# Time-based train-test split

split_index = int(len(X) * 0.80)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("Training data:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting data:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

print("\nTraining period:")
print(X_train.index.min(), "to", X_train.index.max())

print("\nTesting period:")
print(X_test.index.min(), "to", X_test.index.max())
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

rf_model = RandomForestRegressor(
    n_estimators=300,
    random_state=42,
    n_jobs=-1,
    max_features="sqrt"
)

rf_model.fit(X_train, y_train)

y_pred = rf_model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Random Forest Results")
print("---------------------")
print("MAE :", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R²  :", round(r2, 2))
```

```python
def calculate_cpcb_aqi(pm25):

    if pd.isna(pm25):
        return np.nan

    if pm25 <= 30:
        aqi = (50 / 30) * pm25

    elif pm25 <= 60:
        aqi = 50 + ((100 - 50) / (60 - 30)) * (pm25 - 30)

    elif pm25 <= 90:
        aqi = 100 + ((200 - 100) / (90 - 60)) * (pm25 - 60)

    elif pm25 <= 120:
        aqi = 200 + ((300 - 200) / (120 - 90)) * (pm25 - 90)

    elif pm25 <= 250:
        aqi = 300 + ((400 - 300) / (250 - 120)) * (pm25 - 120)

    else:
        aqi = 400 + ((500 - 400) / (500 - 250)) * (pm25 - 250)

    return round(aqi)
```

```python
# ==========================================
# LOG-TRANSFORMED 24-HOUR REGRESSION
# ==========================================

import numpy as np
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


# ==========================================
# TRAINING DATA
# ==========================================

X_train_log = X_train_winter.copy()
X_test_log = X_test_winter.copy()

# Log-transform target
y_train_log = np.log1p(y_train_winter)


# ==========================================
# RANDOM FOREST
# ==========================================

rf_log = RandomForestRegressor(
    n_estimators=300,
    random_state=42,
    n_jobs=-1,
    max_features="sqrt"
)

rf_log.fit(
    X_train_log,
    y_train_log
)

# Prediction is currently in log scale
rf_log_pred = rf_log.predict(X_test_log)

# Convert back to PM2.5 scale
rf_log_pred_pm25 = np.expm1(rf_log_pred)

# Prevent negative predictions
rf_log_pred_pm25 = np.maximum(
    rf_log_pred_pm25,
    0
)


# ==========================================
# XGBOOST
# ==========================================

xgb_log = XGBRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    objective="reg:squarederror"
)

xgb_log.fit(
    X_train_log,
    y_train_log
)

xgb_log_pred = xgb_log.predict(
    X_test_log
)

# Convert back to original PM2.5 scale
xgb_log_pred_pm25 = np.expm1(
    xgb_log_pred
)

# Prevent negative predictions
xgb_log_pred_pm25 = np.maximum(
    xgb_log_pred_pm25,
    0
)


print("===== LOG-TRANSFORMED MODELS TRAINED =====")
print("Random Forest: Done")
print("XGBoost: Done")
```

```python
# ==========================================
# EVALUATE LOG-TRANSFORMED MODELS
# ==========================================

# Random Forest metrics
rf_log_mae = mean_absolute_error(
    y_test_winter,
    rf_log_pred_pm25
)

rf_log_rmse = np.sqrt(
    mean_squared_error(
        y_test_winter,
        rf_log_pred_pm25
    )
)

rf_log_r2 = r2_score(
    y_test_winter,
    rf_log_pred_pm25
)


# XGBoost metrics
xgb_log_mae = mean_absolute_error(
    y_test_winter,
    xgb_log_pred_pm25
)

xgb_log_rmse = np.sqrt(
    mean_squared_error(
        y_test_winter,
        xgb_log_pred_pm25
    )
)

xgb_log_r2 = r2_score(
    y_test_winter,
    xgb_log_pred_pm25
)


# ==========================================
# RESULTS
# ==========================================

print("===== LOG-TRANSFORMED RESULTS =====")

print("\nRandom Forest")
print("MAE :", round(rf_log_mae, 3))
print("RMSE:", round(rf_log_rmse, 3))
print("R²  :", round(rf_log_r2, 3))

print("\nXGBoost")
print("MAE :", round(xgb_log_mae, 3))
print("RMSE:", round(xgb_log_rmse, 3))
print("R²  :", round(xgb_log_r2, 3))
```
